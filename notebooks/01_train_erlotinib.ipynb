# Quick Start: Train PIGE on Erlotinib

This notebook demonstrates how to train PIGE on a single drug (Erlotinib) using the full pipeline.

**Expected runtime:** ~15 minutes

## Pipeline Stages:
1. Pathway Setup: Build pathway interaction network from GO + OmniPath
2. Prep Omics Data: Process mutations, CNA, RNA data
3. Generate PAFE: Create Pathway Activity Feature Engineering matrices
4. Precompute Features: Cache features as tensors
5. Train Models: Train Pathway-based Graph Attention Networks

```python
import sys
from pathlib import Path
import yaml

# Add project root to path
sys.path.insert(0, str(Path.cwd().parent))

from src.pipeline.pipeline_orchestrator import PipelineOrchestrator
from src.main import resolve_config_variables
```

## Setup Configuration
