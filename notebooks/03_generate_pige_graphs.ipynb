- **Pathway nodes**: Colored by importance (red = drives sensitivity, blue = drives resistance)

- **Key genes**: Labeled within pathways as significant targets

## What is a PIGE Graph?
PIGE graphs integrate:
1. Virtual knockout results (from notebook 02)
2. Pathway crosstalk interactions
3. Gene-level importance within pathways
